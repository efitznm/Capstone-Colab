# 1. Community Insights Through Unsupervised Learning

*Using clustering, dimensionality reduction, and exploratory data analysis to uncover patterns in community needs and identify opportunities for intervention.*

### 1. Mapping Community Needs & Resource Gaps
Analyze demographic, socioeconomic, housing, and community-resource data to identify geographic patterns of need.

- Cluster neighborhoods by socioeconomic characteristics.
- Identify areas with overlapping indicators of vulnerability.
- Compare community needs against the geographic distribution of available services.
- Identify potential service deserts.

**Technical Approaches:** K-Means Clustering, Hierarchical Clustering, PCA, Geospatial Analysis.

**Potential Impact:** Help nonprofits and policymakers prioritize where to invest resources and expand services.

### 2. Community Needs Segmentation
Identify distinct profiles of community needs rather than examining each social issue independently.

- Cluster communities or geographic areas by combinations of housing, employment, education, and food insecurity indicators.
- Discover which needs tend to overlap.
- Identify distinct community profiles.

**Technical Approaches:** K-Means, DBSCAN, Hierarchical Clustering, PCA.

**Potential Impact:** Help organizations develop programs that respond to combinations of needs rather than treating every issue in isolation.

### 3. Understanding Patterns in Service Utilization
Explore how people use social services and identify patterns in service engagement.

- Identify common combinations of services accessed.
- Cluster service-utilization patterns.
- Examine referral pathways and potential gaps.
- Identify services that appear underutilized relative to community need.

**Technical Approaches:** Clustering, Association Rule Mining, Network Analysis.

**Potential Impact:** Help organizations understand how people navigate the service system and where coordination could improve.

### 4. Exploring the Characteristics of Housing Instability
Investigate relationships between housing costs, income, employment, transportation, and other socioeconomic indicators.

- Identify clusters of housing vulnerability.
- Explore relationships between contributing factors.
- Compare geographic patterns.
- Identify communities where multiple risk factors intersect.

**Technical Approaches:** Unsupervised Clustering, PCA, Correlation Analysis.

**Potential Impact:** Inform housing program development, community planning, and resource allocation.

# 2. Program Development & Service-Delivery Improvement

*Using data analytics and machine learning to evaluate program effectiveness, identify operational bottlenecks, and improve organizational performance.*

### 5. Nonprofit Service-Delivery Performance Dashboard
Develop a data model and interactive dashboard to monitor nonprofit program performance.

- Track referral-to-service connection rates.
- Measure average time from intake to service.
- Monitor unmet referrals.
- Analyze program enrollment and completion.
- Measure Service-Level Agreement (SLA) compliance.
- Identify service-utilization trends.

**Technical Approaches:** SQL, Python, Data Visualization, KPI Development, Anomaly Detection, Time-Series Analysis.

**AI Extension:** Generate natural-language summaries of performance trends and identify unusual changes in service delivery.

**Potential Impact:** Give nonprofit leadership actionable visibility into organizational performance and opportunities for improvement.

### 6. Transitional Housing Program Outcomes Analysis
Develop an analytical framework to understand factors associated with successful housing transitions.

- Analyze time spent in transitional housing.
- Examine participation in supportive services.
- Explore employment, education, and housing outcomes.
- Identify patterns associated with successful transitions.

**Technical Approaches:** Clustering, Classification (when valid outcome labels exist), Regression, Exploratory Data Analysis.

**Potential Impact:** Help organizations improve program design and understand which supports may be associated with better outcomes.

### 7. Identifying Bottlenecks in Social Service Delivery
Analyze the process from initial intake through referral and service connection.

- Identify where people experience delays.
- Determine where referrals fall through.
- Analyze which stages have the greatest processing times.
- Identify patterns in incomplete service connections.

**Technical Approaches:** Process Mining, Anomaly Detection, Time-Series Analysis, Natural Language Processing (NLP).

**Potential Impact:** Help organizations redesign workflows, improve referral pathways, and increase service accessibility.

# 3. Applied Generative AI & Machine Learning

*Developing practical AI applications that improve resource navigation, support nonprofit decision-making, and strengthen community services.*

### 8. AI-Powered Community Resource Navigator
Develop a system that helps match community needs with available social services.

- Build a structured community resource database.
- Use NLP and semantic search to interpret user requests.
- Match individuals with relevant services.
- Generate personalized resource summaries and next steps.

**Technical Approaches:** NLP, Embeddings, Semantic Similarity, Retrieval-Augmented Generation (RAG), Generative AI.

**Potential Impact:** Reduce the difficulty of navigating fragmented social services and improve access to community resources.

### 9. AI-Assisted Nonprofit Program Evaluation
Create a tool that helps nonprofit staff turn program data into meaningful insights.

- Import program-performance data.
- Calculate key performance indicators.
- Identify trends and anomalies.
- Generate plain-language performance reports.
- Translate findings into potential areas for improvement.

**Technical Approaches:** Generative AI, Anomaly Detection, Statistical Analysis, Data Visualization.

**Potential Impact:** Make data-informed decision-making more accessible to smaller nonprofits without dedicated data teams.

### 10. Understanding Community Feedback Through NLP
Analyze qualitative feedback from surveys, public comments, or appropriately de-identified program evaluations.

- Identify recurring themes.
- Cluster common concerns.
- Explore sentiment and changes over time.
- Compare community feedback with existing services.

**Technical Approaches:** NLP, Topic Modeling, Embeddings, Sentiment Analysis.

**Potential Impact:** Help organizations incorporate community voices into program development and strategic decision-making.

## Overall Considerations

When selecting a capstone project, consider:

- **Feasibility:** Can the project be completed within the capstone timeline?
- **Data availability:** Can we access enough relevant, reliable data?
- **Technical depth:** Does the project demonstrate the data science and machine learning skills we've learned?
- **Practical application:** Can the results inform a real decision or solve a meaningful problem?
- **Scalability:** Could the project be expanded or developed further after the bootcamp?

### A Final Question

**Are we building a model simply to demonstrate a technical skill, or are we using that skill to generate insights that can lead to meaningful action?**

The strongest capstone projects ideally do both.
